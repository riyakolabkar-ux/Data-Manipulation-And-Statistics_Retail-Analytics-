```python
# Task 1: Understanding Skewness
# Import Required Libraries

import pandas as pd
import numpy as np
```

```python
#Load Dataset Using Pandas
df = pd.read_csv("/content/Retail_Dataset_Cleaned 8.csv")

print("Dataset Loaded Successfully")
```

```python
# Display First Five Records

df.head()
```

```python
# Check Dataset Information

df.info()
```

```python
# Display Column Names

df.columns
```

```python
#Select Revenue Column

df["Revenue"]
```

```python
# Calculate Revenue Skewness

revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
# Interpret Revenue Skewness

if revenue_skew > 0:
    print("Positive Skew")
elif revenue_skew < 0:
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

```python
# Calculate Units Sold Skewness

units_skew = df["Units_Sold"].skew()
print(units_skew)
```

```python
# Task 2: Understanding Distribution Patterns
# Import Visualization Library

import matplotlib.pyplot as plt
```

```python
# Plot Revenue Distribution

plt.hist(df["Revenue"], bins=20)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
from matplotlib.typing import LineStyleType
# Plot Units Sold Distribution

plt.hist(df["Units_Sold"], bins=20, color="orange")
plt.title("Units Sold Distribution for Retail Dataset",fontweight="bold",color="red")
plt.xlabel("Units Sold",fontweight="bold",color="red")
plt.ylabel("Frequency",fontweight="bold",color="red")
plt.ylabel("frequency")
plt.grid(axis="y",linestyle="--",color="black",alpha=0.2)
plt.show()
```

```python
# Display Final Summary Table

summary = pd.DataFrame({
    "Metric":["Revenue","Units Sold"],
    "Skewness":[revenue_skew,units_skew]
})
summary
```
