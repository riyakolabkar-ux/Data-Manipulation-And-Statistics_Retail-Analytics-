```python
# Import Required Libraries

import pandas as pd
import numpy as np
```

```python
# Load Dataset Using Pandas

df = pd.read_csv("/content/Retail_Dataset_Cleaned 10.csv")

print("Dataset Loaded Successfully")
```

```python
# Display First Five Records

df.head()
```

```python
# Check Dataset Information
df.info()
```

```python
# Display Column Names

df.columns
```

```python
# Create Category-wise Revenue Pivot Table

category_revenue = pd.pivot_table(
    df,
    values="Revenue",
    index="Category",
    aggfunc="sum"
)

category_revenue
```

```python
# Create City-wise Units Sold Pivot Table

city_sales = pd.pivot_table(
    df,
    values="Units_Sold",
    index="City",
    aggfunc="sum"
)
city_sales
```

```python
# Create Customer Segment-wise Revenue Report

segment_revenue = pd.pivot_table(
    df,
    values="Revenue",
    index="Customer_Segment",
    aggfunc="sum"
)

segment_revenue
```

```python
# Task 2: Aggregating Retail Data for Business Decision-Making
# Calculate Multiple Revenue Statistics Category-wise

category_summary = pd.pivot_table(
    df,
    values="Revenue",
    index="Category",
    aggfunc=["sum", "mean", "max", "min"]
)

category_summary
```

```python
# Aggregate Units Sold by Sales Channel

channel_summary = pd.pivot_table(
    df,
    values="Units_Sold",
    index="Sales_Channel",
    aggfunc=["sum", "mean"]
)

channel_summary
```

```python
# Aggregate Revenue and Profit Margin by City

city_performance = pd.pivot_table(
    df,
    values=["Revenue", "Profit_Margin"],
    index="City",
    aggfunc={
        "Revenue": "sum",
        "Profit_Margin": "mean"
    }
)
city_performance
```

```python
# Create Store Type Performance Report

store_summary = pd.pivot_table(
    df,
    values=["Revenue", "Units_Sold"],
    index="Store_Type",
    aggfunc="sum"
)

store_summary
```

```python
# Category wise minimum unit sold,maximum revenue and total profit margin
category_revenue = pd.pivot_table(
    df,
    values=["Revenue", "Units_Sold" , "Profit_Margin"],
    index="Category",
    aggfunc={
        "Revenue": "max",
        "Units_Sold": "min",
        "Profit_Margin": "sum"
    }
)
category_revenue
```
